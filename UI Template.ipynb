```python
import customtkinter, tkinter

#root = Tk()

customtkinter.set_appearance_mode("dark")

#Change the color of the button
customtkinter.set_default_color_theme("blue")

root = customtkinter.CTk()

root.geometry('400x500')

mybutton = customtkinter.CTkButton(master=root,text= "Hello World!")

mybutton.place(relx = .5, rely=.5, anchor = CENTER)


# create scrollable textbox
tk_textbox = tkinter.Text(root, highlightthickness=0)
tk_textbox.grid(row=0, column=0, sticky="nsew")

# create CTk scrollbar
ctk_textbox_scrollbar = customtkinter.CTkScrollbar(root, command=tk_textbox.yview)
ctk_textbox_scrollbar.grid(row=0, column=3, sticky="ns")

root.mainloop()
```
